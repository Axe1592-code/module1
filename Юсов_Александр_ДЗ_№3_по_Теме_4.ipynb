Задание 1


---

Переведите содержимое файла purchase_log.txt в словарь purchases вида:
* {‘1840e0b9d4’: ‘Продукты’, …}

Пример работы программы при выводе первых двух элементов словаря purchases:

* 1840e0b9d4 ‘Продукты‘
* 4e4f90fcfb ‘Электроника‘

---


```python
"""Чтение purchase_log.txt и построение словаря purchases вида {user_id: category}"""
import json

purchases = {}

with open('purchase_log.txt') as f:
    next(f)  # Пропуск первой строки ({"user_id": "user_id", "category": "category"})
    for line in f:
        line = line.strip()
        dict_ = json.loads(line)
        purchases[dict_['user_id']] = dict_['category']

# Вывод первых двух элементов словаря [:2]
for user_id, category in list(purchases.items())[:2]:
    print(user_id, f"'{category}'")
```

Задание 2


---
Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки.

Если покупка была, сам файл visit_log.csv изменять не надо.

Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.
Учтите условия для данных:

* Содержимое purchase_log.txt помещается в оперативную память компьютера;

* Содержимое visit_log.csv — нет, используйте только построчную обработку этого файла.

---

```python
"""Построчное чтение visit_log.csv и добавление категории из purchases с записью обогащенных визитов в funnel.csv"""
import csv
from pathlib import Path

visit_log_path = Path('visit_log.csv')
funnel_path = Path('funnel.csv')

# Открытие visit_log.csv для чтения и funnel.csv для записи
with open(visit_log_path, 'r', encoding='utf-8') as in_, \
     open(funnel_path, 'w', encoding='utf-8', newline='') as out_:

    read = csv.reader(in_)
    write = csv.writer(out_)

    # Заголовок нового файла
    write.writerow(['user_id', 'source', 'category'])

    # Пропуск заголовка visit_log.csv ('user_id', 'source')
    next(read)

    # Построчная обработка
    for row in read:
        user_id = row[0]
        source = row[1]
        if user_id in purchases:
            write.writerow([user_id, source, purchases[user_id]])

# Проверка (первые 3 строки funnel.csv)
with open(funnel_path) as f:
    print([next(f).strip() for x in range(3)])
```
