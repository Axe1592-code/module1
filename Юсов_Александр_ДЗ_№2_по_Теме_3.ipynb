Напишите код на Python для решения следующей задачи.

Реализуйте функцию sum_distance(from, to), которая суммирует все числа от значения from до величины to включительно.

Примечание. Если пользователь задаст первое число, которое окажется больше второго, просто поменяйте их местами.

---



```python
# Программа 1 (Через print)
def sum_distance(from_, to):
    """Суммирует все числа от значения from_ до величины to включительно"""
    # Если первое число больше второго — меняем местами
    if from_ > to:
        from_, to = to, from_
    total = 0
    for number in range(from_, to + 1):
        total += number
    return total

print(sum_distance(1, 5))
print(sum_distance(5, 1))
print(sum_distance(3, 3))
```

```python
# Программа 2 (Через input)
def sum_distance(from_, to):
    """Суммирует все числа от значения from_ до величины to включительно"""
    # Если первое число больше второго — меняем местами
    if from_ > to:
        from_, to = to, from_
    total = 0
    for number in range(from_, to + 1):
        total += number
    return total

# Ввод данных с преобразованием из строки в целое число (так код компактнее получается)
from_ = int(input("Введи первое число: "))
to = int(input("Введи второе число: "))
print("Сумма чисел:", sum_distance(from_, to))
```

Задание 2


---

Напишите код на Python для решения следующей задачи.

Реализуйте функцию trim_and_repeat(), которая принимает три параметра:

*   строку;
*   offset — число символов, на которое нужно обрезать строку слева;
*   repetitions — сколько раз нужно повторить строку перед возвратом получившейся строки.

Число символов для среза по умолчанию равно 0, а количество повторений — 1.

Функция должна возвращать полученную строку.

---

```python
# Программа 1 (Через print)
def trim_and_repeat(string, offset=0, repetitions=1):
    """Обрезает строку слева на N символов и повторяет строку перед возвратом получившейся строки"""
    return string[offset:] * repetitions

print(trim_and_repeat("Python"))
print(trim_and_repeat("Python", 2))
print(trim_and_repeat("Python", 2, 3))
```

```python
# Программа 2 (Через input)
def trim_and_repeat(string, offset=0, repetitions=1):
    """Обрезает строку слева на N символов и повторяет строку перед возвратом получившейся строки"""
    return string[offset:] * repetitions

string = input("Строка: ")
offset = int(input("Offset: "))
repetitions = int(input("Repetitions: "))
print("Результат:", trim_and_repeat(string, offset, repetitions))
```
